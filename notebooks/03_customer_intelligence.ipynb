# 03. Customer Intelligence & RFP Analysis

Computes user-level lifetime order frequency, cadence intervals, basket intensity, and quantile RFP scores.

```python
import duckdb
import pandas as pd
from pathlib import Path

con = duckdb.connect()
cust_pq = Path('../data/features/customer_features.parquet')
df = con.execute(f"SELECT * FROM read_parquet('{cust_pq.as_posix()}') LIMIT 20").fetchdf()
print(df[['user_id', 'total_orders', 'avg_basket_size', 'reorder_rate', 'rfp_segment']])
```
