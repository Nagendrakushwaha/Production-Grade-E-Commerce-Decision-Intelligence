# 02. Exploratory Data Analysis (EDA)

Examines distribution of order frequency, hour of day, day of week, basket sizes, and category volumes.

```python
import duckdb
import pandas as pd
from pathlib import Path

con = duckdb.connect()
pq_dir = Path('../data/parquet')

con.execute(f"CREATE OR REPLACE VIEW orders AS SELECT * FROM read_parquet('{(pq_dir / 'orders.parquet').as_posix()}')")
dow_df = con.execute("SELECT order_dow, count(*) as orders FROM orders GROUP BY order_dow ORDER BY order_dow").fetchdf()
print(dow_df)
```

```python
hour_df = con.execute("SELECT order_hour_of_day, count(*) as orders FROM orders GROUP BY order_hour_of_day ORDER BY order_hour_of_day").fetchdf()
print(hour_df.head(10))
```
